```python
tenk = apple.get_filings(form="10-K").latest()
xbrl = tenk.xbrl()
bs = xbrl.statements.balance_sheet()
bs_df = bs.to_dataframe()
```

```python
assets = bs_df.loc[
    bs_df["standard_concept"] == "Assets",
    "2025-09-27"
].iloc[0]

liabilities = bs_df.loc[
    bs_df["standard_concept"] == "Liabilities",
    "2025-09-27"
].iloc[0]

equity = bs_df.loc[
    bs_df["standard_concept"] == "AllEquityBalance",
    "2025-09-27"
].iloc[0]

liab_equity = bs_df.loc[
    bs_df["standard_concept"] == "LiabilitiesAndEquity",
    "2025-09-27"
].iloc[0]

print("Assets:", assets)
print("Liabilities:", liabilities)
print("Equity:", equity)
print("Liabilities + Equity:", liab_equity)
```

```python
residual_1 = assets - (liabilities + equity)
residual_2 = assets - liab_equity

print("Assets - (Liabilities + Equity):", residual_1)
print("Assets - reported LiabilitiesAndEquity:", residual_2)
```
